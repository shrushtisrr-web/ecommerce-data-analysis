```python
import matplotlib.pyplot as plt
#sales by cateory
category_sales = df.groupby('Category')['Amount'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
category_sales.plot(kind='bar')

plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
monthly_sales = df.groupby(df['new_date'].dt.month)['Amount'].sum()

plt.figure(figsize=(10,5))
monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(range(1, 13))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
status_counts = df['Status'].value_counts()

plt.figure(figsize=(10,6))
bars = status_counts.plot(kind='bar')

plt.title('Order Status Distribution')
plt.xlabel('Order Status')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45, ha='right')

# Add numbers on top of bars
for bar in bars.patches:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f'{int(height):,}',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.show()
```

```python
fulfilment_sales = df.groupby('Fulfilment')['Amount'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
bars = fulfilment_sales.plot(kind='bar')

plt.title('Sales by Fulfilment')
plt.xlabel('Fulfilment')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)

# Add values on bars
for bar in bars.patches:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f'{height:,.0f}',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.show()
#showing which fulfillment has more sales
```

```python
state_sales = (
    df.groupby('ship-state')['Amount']
    .sum()
    .sort_values(ascending=False)
    .head(7)
)

plt.figure(figsize=(9,9))

plt.pie(
    state_sales.values,
    labels=state_sales.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Sales Distribution — Top 7 States', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
city_sales = (
    df.groupby('ship-city')['Amount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,6))

bars = plt.bar(
    city_sales.index,
    city_sales.values,
    color='orange'
)

plt.title('Top 10 Cities by Sales', fontsize=16)
plt.xlabel('City')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right')

# Add values on bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f'{height:,.0f}',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.show()
```

```python
category_sales = (
    df.groupby('Category')['Amount']
    .sum()
    .sort_values(ascending=True)
)

plt.figure(figsize=(10,6))

bars = plt.barh(
    category_sales.index,
    category_sales.values,
    color='teal'
)

plt.title('Sales by Category', fontsize=16)
plt.xlabel('Total Sales')
plt.ylabel('Category')

# Add values
for bar in bars:
    width = bar.get_width()
    plt.text(
        width,
        bar.get_y() + bar.get_height()/2,
        f'₹{width:,.0f}',
        va='center',
        ha='left'
    )

plt.tight_layout()
plt.show()
```

key insights found in the data 

1. Category performance
Set generated the highest sales: ₹39.20M
Kurta was second: ₹21.30M
Western Dress was third: ₹11.22M
Set and Kurta also had very similar quantities sold.

Business insight: Sales are heavily concentrated in a few major product categories, especially Set and Kurta.

2. Order status

The largest status groups were:

Shipped: 77,804
Shipped - Delivered to Buyer: 28,769
Cancelled: 18,332

There are also smaller groups involving returns, pending orders, delivery issues, etc.

Business insight: Most orders are in the shipping/delivery pipeline, but cancellations are large enough to deserve attention.

3. Quantity / data-quality finding
12,807 rows have Qty = 0.
Of these, 12,701 are Cancelled.
However, 5,136 zero-quantity rows still contain an Amount, so we correctly didn't simply delete them.

Business insight: Quantity and order status need to be interpreted together; zero quantity doesn't automatically mean the row is invalid.

4. Fulfilment

Our analysis showed Amazon fulfilment has higher sales than Merchant fulfilment.

Business insight: Amazon fulfilment is the larger contributor to sales in this dataset.

5. Customer geography

We analysed:

Top states by sales
Top cities by sales

These can help identify geographic markets where sales are concentrated.

6. Sales trend

We also created the monthly sales trend, which we'll use later in Power BI for a more polished interactive version.

7. Data quality
Duplicates: 0
Amount has 7,795 missing values
These were not blindly converted to zero because many correspond to cancelled/no-sale records.
We also created the Data_Quality_Flag during Excel cleaning.
